```python
#@title Play "MadLibs" here! (Veer's version)
Your_Name = "Veer" #@param {type:"string"}

def passtheword():
  password = None
  attempts = []
  while password != "madlibs":
    password = input ("Enter the password: ")
    attempts.append(password)
    if password != "madlibs":
      print("Your password is incorrect. You have guessed:")
      for i in attempts:
        print (i)
  print("You said Mad Libs. Let's play it then!")
#
#
def madlibs():
  words = []
  def prompt(kind):
    ask = ("Please enter " + kind + ": ")
    word = input(ask)
    words.append(word)

  prompt("an adjective")

  prompt("a nationity")

  prompt("a name for a person")

  prompt("an plural noun")

  prompt("an adjective")

  prompt("an plural noun")

  print("Computers were invented by a", words[0], words[1], "engineer named", words[2],". To make a computer, you need to take a lot of", words[3], ", melt them down, and make", words[4], words[5], "." )
#
#
#
#
#
passtheword()
madlibs()
print("Thank you for playing, " + Your_Name + ".")
```
